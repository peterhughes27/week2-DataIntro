## 1. Research Question

- **Main research question:** How accurately can concrete compressive strength be predicted from mixture composition and curing age?
- **Why is this question interesting or important?** This question is important because predicting concrete compressive strength can help estimate whether a concrete mix will meet required performance levels before extensive physical testing is completed. It is also interesting because strength depends on the combined effects of several mixture components and curing age rather than one single factor.
- **Who might use or care about the answer?** Civil and structural engineers, concrete manufacturers, construction companies, and materials researchers could all use this information. It could help them assess concrete mix designs, estimate expected strength, and identify which mixture factors are most important before or alongside physical testing.
- **Possible sub questions** "Which mixture components have the strongest relationship with compressive strength?", "Which machine-learning model gives the most accurate predictions for compressive strength?"

## Research Questions and Answers

For every answer, distinguish between **known evidence**, **reasonable interpretation**, and **unknown information to investigate**.

### 1. Where did the data come from?

**Answer:** Known evidence: The dataset is the Concrete Compressive Strength dataset created by I-Cheng Yeh and is hosted by the UCI Machine Learning Repository. It contains 1,030 concrete samples, with eight input variables describing mixture composition and curing age, and one output variable giving compressive strength in MPa. The UCI page cites Yeh's 1998 paper Modeling of Strength of High-Performance Concrete Using Artificial Neural Networks as the introductory paper and gives the dataset DOI as 10.24432/C5PK67.

Reasonable interpretation: The data appear to come from laboratory concrete mix experiments used to study how different mix proportions and curing age affect compressive strength. Yeh's paper states that trial batches of high-performance concrete were produced in the laboratory and tested experimentally.

Unknown information to investigate: The UCI page does not clearly state the exact laboratory location or full sampling procedure for all 1,030 observations, so this would need further investigation if required.

### 2. Why was it originally collected?

**Answer:** Known evidence: The data were originally used to investigate and model the relationship between concrete mixture proportions, curing age and compressive strength. Yeh's 1998 study specifically aimed to test whether artificial neural networks could accurately predict the compressive strength of high-performance concrete.

Reasonable interpretation: The dataset was therefore collected to improve understanding and prediction of concrete strength where the relationship between ingredients and strength is nonlinear and difficult to describe using a simple formula.

Unknown information to investigate: It is not fully clear from the UCI summary whether every row in the current dataset came from the exact same experimental campaign described in the 1998 paper or whether some observations were later compiled.

For our current project, the purpose is very similar to the original one: we are using the same variables to predict compressive strength, so there is no major change in purpose and no obvious ethical concern.

### 3. Who is represented?

**Answer:** Known evidence: The dataset does not represent people or human populations. Each row represents a concrete mixture/sample, described by cement, blast-furnace slag, fly ash, water, superplasticiser, coarse aggregate, fine aggregate and curing age, with measured compressive strength as the target.

Reasonable interpretation: The represented population is therefore a set of laboratory-tested concrete mixtures, mainly relevant to high-performance concrete within the range of compositions and ages included in the dataset.

Unknown information to investigate: The exact geographic origin of all materials, the precise laboratory location, and whether the samples represent commercial mixes or only experimental formulations are not fully documented on the UCI dataset page.

### 4. Who might be missing?

**Answer:** Known evidence: The dataset contains concrete mixtures rather than people, so there are no missing demographic groups or non-responses in the usual sense.

Reasonable interpretation: Some types of concrete may be under-represented or absent, such as mixtures using different cement types, alternative aggregates, fibres, unusual admixtures, or very different curing conditions. This means the dataset may not represent every concrete formulation used in real construction.

Unknown information to investigate: The source does not fully document whether the samples cover all major commercial concrete types or how representative they are of concrete used in different countries and environments.

### 5. What personal or sensitive information does it contain, explicitly or implicitly?

**Answer:** Known evidence: The dataset contains no personal or sensitive information. It records only material properties and experimental measurements such as cement content, water content, aggregates, curing age and compressive strength.

There are no names, addresses, identifiers, demographic variables or other fields that could identify an individual. Therefore, there are no meaningful privacy or re-identification concerns.

### 6. What could your analysis reveal that was not obvious in the original data?

**Answer:** Known evidence: The raw dataset provides the individual mixture measurements and their measured compressive strengths.

Reasonable interpretation: Our analysis could reveal which variables are most strongly associated with compressive strength, how several ingredients interact, and whether machine-learning models can predict strength accurately for unseen mixtures. It may also reveal nonlinear relationships that are difficult to identify by looking at the raw values alone.

For example, the analysis may show that curing age is important but that its effect depends on the mixture composition, or that certain combinations of water, cement and supplementary materials lead to stronger concrete.

Unknown information to investigate: The exact relationships and the most important predictors cannot be known until the exploratory analysis and model evaluation are completed.

### 7. Who could benefit from your findings?

**Answer:** Civil and structural engineers, concrete manufacturers, construction companies and materials researchers could benefit from the findings. A reliable prediction model could help them estimate expected compressive strength from a proposed mixture, compare different mix designs, and identify promising formulations before carrying out full physical strength testing.

The results could therefore support mix optimisation, quality control and more efficient material testing, although physical testing would still be required for real-world engineering decisions.

### 8. Who could be harmed or misrepresented?

**Answer:** Known evidence: The dataset contains no personal information, so there is little risk of privacy loss, discrimination or stigma towards individuals or demographic groups.

Reasonable interpretation: The main risk would come from misusing or over-trusting the model's predictions. Engineers, construction companies, building owners or users of structures could potentially be affected if predicted compressive strength was treated as a substitute for proper physical testing and engineering standards.
